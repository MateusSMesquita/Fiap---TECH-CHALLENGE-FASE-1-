## IA na Saúde da Mulher: Classificação e Predição de Risco para Câncer de Mama

Tech Challenge

Machine Learning aplicado à saúde feminina

Objetivo: desenvolver e avaliar modelos de Machine Learning capazes de classificar casos de câncer de mama como benignos ou malignos, utilizando dados clínicos e características extraídas de exames, com foco em apoio à decisão médica.

Tecnologias: Python · Pandas · NumPy · Scikit-learn · Matplotlib · Seaborn · SHAP · Google ColabTech Challenge —

O câncer de mama representa um importante desafio para a saúde pública, tornando fundamental a identificação precoce de alterações suspeitas. Neste projeto, será desenvolvida uma solução baseada em Machine Learning capaz de analisar características associadas a exames de mama e classificar os casos entre benignos e malignos.

A proposta não é substituir a avaliação médica, mas demonstrar como técnicas de Inteligência Artificial podem ser utilizadas como ferramentas de apoio à análise e à tomada de decisão.

O dataset Breast Cancer Wisconsin Diagnostic que foi escolhido não contém as imagens dos exames. Ele contém dados numéricos extraídos de imagens de massas mamárias. Portanto, vamos treinar modelos para classificar os casos usando essas características numéricas.


Bibliotecas utilizadas
*   Pandas: utilizada para carregar, organizar, manipular e analisar os dados do dataset.

*   NumPy: utilizada para operações numéricas e manipulação de arrays.


*   Matplotlib: utilizada para criação de gráficos e visualizações dos resultados.
*   Seaborn: utilizada para visualizações estatísticas, como distribuições e matriz de correlação.



*   Scikit-learn: principal biblioteca de Machine Learning do projeto, utilizada no pré-processamento, treinamento, classificação e avaliação dos modelos.
*   SHAP: utilizada para interpretar as previsões dos modelos e identificar quais características mais influenciaram cada resultado.

Em conjunto, essas bibliotecas permitem realizar todo o fluxo do projeto: exploração dos dados, pré-processamento, treinamento, avaliação e explicabilidade dos modelos de classificação entre casos benignos (B) e malignos (M).

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

import shap
```

```python
exames["diagnosis"] = exames["diagnosis"].map({
    "B": 0,
    "M": 1
    })
```

```python
exames["diagnosis"].value_counts()
```

```python
X = exames.drop(columns=["diagnosis"])
y = exames["diagnosis"]
```

```python
print("X:", X.shape)
print("y:", y.shape)
```

```python
SEED = 42
np.random.seed(SEED)
```

```python
X_temp, X_test, y_temp, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=SEED,
    stratify=y
)
```

```python
print("Total de registros:", len(X))
print("X_temp:", len(X_temp))
print("X_test:", len(X_test))

print("\nPercentuais:")
print("X_temp:", len(X_temp) / len(X) * 100, "%")
print("X_test:", len(X_test) / len(X) * 100, "%")
```

```python
X_train, X_val, y_train, y_val = train_test_split(
    X_temp,
    y_temp,
    test_size=0.125,
    random_state=SEED,
    stratify=y_temp
)
```

```python
print("Treino:", X_train.shape)
print("Validação:", X_val.shape)
print("Teste:", X_test.shape)
```

```python
print("\nPercentuais:")
print("Treino:", len(X_train) / len(X) * 100, "%")
print("Validação:", len(X_val) / len(X) * 100, "%")
print("Teste:", len(X_test) / len(X) * 100, "%")
```

```python
print("TREINO")
print(y_train.value_counts())

print("\nVALIDAÇÃO")
print(y_val.value_counts())

print("\nTESTE")
print(y_test.value_counts())
```

```python
modelo_logistico = Pipeline([
    ("scaler", StandardScaler()),
    ("modelo", LogisticRegression(
        random_state=SEED,
        max_iter=1000
    ))
])
```

```python
modelo_rf = RandomForestClassifier(
    n_estimators=100,
    random_state=SEED
)
```

```python
print("Iniciando treinamento da Regressão Logística...")

modelo_logistico.fit(X_train, y_train)

print("Regressão Logística treinada com sucesso!")
```

```python
print("Iniciando treinamento do Random Forest...")

modelo_rf.fit(X_train, y_train)

print("Random Forest treinado com sucesso!")
```

```python
print("Coeficientes da Regressão Logística:")
print(modelo_logistico.named_steps["modelo"].coef_.shape)
```

```python
print("Quantidade de árvores no Random Forest:")
print(len(modelo_rf.estimators_))
```

```python
y_pred_logistico = modelo_logistico.predict(X_val)
y_pred_rf = modelo_rf.predict(X_val)
```

```python
print("Previsões - Regressão Logística:")
print(y_pred_logistico)

print("\nPrevisões - Random Forest:")
print(y_pred_rf)
```

```python
print("Quantidade de características:", X.shape[1])
print("\nColunas utilizadas:")
print(X.columns.tolist())
```

```python
print("=== REGRESSÃO LOGÍSTICA ===")

print("Accuracy:",
      accuracy_score(y_val, y_pred_logistico))

print("Precision:",
      precision_score(y_val, y_pred_logistico))

print("Recall:",
      recall_score(y_val, y_pred_logistico))

print("F1-score:",
      f1_score(y_val, y_pred_logistico))


print("\n=== RANDOM FOREST ===")

print("Accuracy:",
      accuracy_score(y_val, y_pred_rf))

print("Precision:",
      precision_score(y_val, y_pred_rf))

print("Recall:",
      recall_score(y_val, y_pred_rf))

print("F1-score:",
      f1_score(y_val, y_pred_rf))
```

```python
cm_logistico = confusion_matrix(y_val, y_pred_logistico)
cm_rf = confusion_matrix(y_val, y_pred_rf)

print("Matriz de Confusão - Regressão Logística")
print(cm_logistico)

print("\nMatriz de Confusão - Random Forest")
print(cm_rf)
```

```python
y_pred_logistico_test = modelo_logistico.predict(X_test)

print("=== RESULTADOS NO CONJUNTO DE TESTE ===")

print("Accuracy:",
      accuracy_score(y_test, y_pred_logistico_test))

print("Precision:",
      precision_score(y_test, y_pred_logistico_test))

print("Recall:",
      recall_score(y_test, y_pred_logistico_test))

print("F1-score:",
      f1_score(y_test, y_pred_logistico_test))
```

```python
cm_test = confusion_matrix(y_test, y_pred_logistico_test)

print(cm_test)
```

O modelo apresentou desempenho elevado tanto na validação quanto no conjunto de teste. Entretanto, houve redução do Recall de 100% na validação para 92,86% no teste, indicando que o desempenho observado na validação não se manteve exatamente no conjunto de dados não utilizado durante o desenvolvimento. No teste, foram identificados corretamente 39 dos 42 casos malignos, enquanto 3 foram classificados como benignos.

```python
coeficientes = modelo_logistico.named_steps["modelo"].coef_[0]

importancias_logistica = pd.Series(
    abs(coeficientes),
    index=X.columns
).sort_values(ascending=False)

print("Top 10 características mais importantes:")
print(importancias_logistica.head(10))
```

```python
plt.figure(figsize=(10, 6))

importancias_logistica.head(10).sort_values().plot(kind="barh")

plt.title("Top 10 Características - Regressão Logística")
plt.xlabel("Importância absoluta do coeficiente")
plt.ylabel("Característica")

plt.show()
```

```python
modelo = modelo_logistico.named_steps["modelo"]
scaler = modelo_logistico.named_steps["scaler"]

X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
explainer = shap.LinearExplainer(
    modelo,
    X_train_scaled
)

shap_values = explainer(X_test_scaled)
```

```python
shap.plots.beeswarm(
    shap_values,
    max_display=10
)
```

```python

X_train_scaled_df = pd.DataFrame(
    X_train_scaled,
    columns=X.columns,
    index=X_train.index
)

X_test_scaled_df = pd.DataFrame(
    X_test_scaled,
    columns=X.columns,
    index=X_test.index
)

explainer = shap.LinearExplainer(
    modelo,
    X_train_scaled_df
)

shap_values = explainer(X_test_scaled_df)

shap.plots.beeswarm(
    shap_values,
    max_display=10
)
```

```python
coeficientes = modelo_logistico.named_steps["modelo"].coef_[0]

importancias_logistica = pd.Series(
    abs(coeficientes),
    index=X.columns
).sort_values(ascending=False)

print("Top 10 características mais importantes:")
print(importancias_logistica.head(10))
```

```python
plt.figure(figsize=(10, 6))

importancias_logistica.head(10).sort_values().plot(kind="barh")

plt.title("Top 10 Características - Regressão Logística")
plt.xlabel("Importância absoluta do coeficiente")
plt.ylabel("Característica")

plt.show()
```

```python
print("=== TESTE - REGRESSÃO LOGÍSTICA ===")

y_pred_logistico_teste = modelo_logistico.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred_logistico_teste))
print("Precision:", precision_score(y_test, y_pred_logistico_teste))
print("Recall:", recall_score(y_test, y_pred_logistico_teste))
print("F1-score:", f1_score(y_test, y_pred_logistico_teste))
print("Matriz de Confusão:")
print(confusion_matrix(y_test, y_pred_logistico_teste))


print("\n=== TESTE - RANDOM FOREST ===")

y_pred_rf_teste = modelo_rf.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred_rf_teste))
print("Precision:", precision_score(y_test, y_pred_rf_teste))
print("Recall:", recall_score(y_test, y_pred_rf_teste))
print("F1-score:", f1_score(y_test, y_pred_rf_teste))
print("Matriz de Confusão:")
print(confusion_matrix(y_test, y_pred_rf_teste))
```

```python
importancias_rf = pd.Series(
    modelo_rf.feature_importances_,
    index=X.columns
).sort_values(ascending=False)

print("Top 10 características mais importantes - Random Forest:")
print(importancias_rf.head(10))
```

```python
plt.figure(figsize=(10, 6))

importancias_rf.head(10).sort_values().plot(kind="barh")

plt.title("Top 10 Características - Random Forest")
plt.xlabel("Importância")
plt.ylabel("Característica")

plt.show()
```

```python
explainer_rf = shap.TreeExplainer(modelo_rf)

shap_values_rf = explainer_rf.shap_values(X_test)

shap.summary_plot(
    shap_values_rf,
    X_test,
    max_display=10
)
```

```python

explainer_rf = shap.TreeExplainer(modelo_rf)


shap_values_rf = explainer_rf.shap_values(X_test)

# Gráfico SHAP
shap.summary_plot(
    shap_values_rf,
    X_test,
    max_display=10
)
```

## Conclusão

Foram desenvolvidos e avaliados dois modelos de classificação utilizando a base Wisconsin Diagnostic Breast Cancer (WDBC): Regressão Logística e Random Forest.

Os dois modelos apresentaram desempenho elevado. No conjunto de teste, o Random Forest apresentou o melhor desempenho geral, alcançando 97,37% de acurácia, 100% de precision, 92,86% de recall e 96,30% de F1-score. Apesar da vantagem, a diferença em relação à Regressão Logística foi pequena, e ambos apresentaram o mesmo recall de 92,86%.

As técnicas de Feature Importance e SHAP foram utilizadas para analisar quais características tiveram maior influência nas previsões dos modelos, contribuindo para uma melhor interpretação dos resultados.

Apesar do desempenho obtido, o modelo não deve ser utilizado de forma autônoma para realizar diagnósticos médicos. Os resultados foram obtidos a partir de uma base de dados específica e, em uma aplicação real, seriam necessárias validações adicionais. O modelo pode ser considerado uma ferramenta de apoio à análise e à identificação de padrões, mantendo a decisão final sob responsabilidade do profissional de saúde.
